```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
def circle(size):
    
    array = np.zeros((size, size), dtype=int)
    center_x, center_y = size/2, size/2
    radius = size/4

    # Create a mesh grid for the entire array
    x, y = np.meshgrid(np.arange(size), np.arange(size))

    # Use the circle equation to set values within the circle to 1
    inside_circle = (x - center_x)**2 + (y - center_y)**2 <= radius**2
    array[inside_circle] = 1
    
    return array
```

```python
fig, ax = plt.subplots(1,3,dpi=300,figsize=(10,30))



ax[0].imshow(circle(1000), cmap='binary', interpolation=None)
ax[1].imshow(circle(50), cmap='binary', interpolation=None)

arr = circle(50)
im = ax[2].imshow(circle(50), cmap='binary', interpolation=None)
for i in range(50):
    for j in range(50):
        ax[2].text(j, i, str(arr[i, j]), ha='center', va='center', color='red', fontsize=4.5)

ax[0].axis('off')
ax[1].axis('off')
ax[2].axis('off')

plt.tight_layout()
plt.show()
```
